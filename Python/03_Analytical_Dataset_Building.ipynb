```python
# ============================================================
# STEP 2 — VALIDATE THE BASE ENCOUNTERS DATASET
# ============================================================

print("Checking the base encounters dataset...")
print()


# ------------------------------------------------------------
# 1. CHECK DUPLICATE ENCOUNTER IDs
# ------------------------------------------------------------

duplicate_ids = encounters["Id"].duplicated().sum()

print("Duplicate Encounter IDs:", duplicate_ids)


# ------------------------------------------------------------
# 2. CHECK DUPLICATE COMPLETE ROWS
# ------------------------------------------------------------

duplicate_rows = encounters.duplicated().sum()

print("Duplicate Complete Rows:", duplicate_rows)


# ------------------------------------------------------------
# 3. CHECK MISSING ENCOUNTER IDs
# ------------------------------------------------------------

missing_ids = encounters["Id"].isna().sum()

print("Missing Encounter IDs:", missing_ids)


# ------------------------------------------------------------
# 4. CHECK MISSING TOTAL CLAIM COST
# ------------------------------------------------------------

missing_cost = encounters["TOTAL_CLAIM_COST"].isna().sum()

print("Missing TOTAL_CLAIM_COST:", missing_cost)


# ------------------------------------------------------------
# 5. CHECK NEGATIVE TOTAL CLAIM COST
# ------------------------------------------------------------

negative_cost = (encounters["TOTAL_CLAIM_COST"] < 0).sum()

print("Negative TOTAL_CLAIM_COST:", negative_cost)


# ------------------------------------------------------------
# 6. FINAL SUMMARY
# ------------------------------------------------------------

print()
print("Total Encounters:", len(encounters))

print()
print("Base dataset validation completed.")
```

```python
# ============================================================
# STEP 3 — LOAD AND VALIDATE PATIENTS DATASET
# ============================================================

# ------------------------------------------------------------
# 1. LOAD PATIENTS DATASET
# ------------------------------------------------------------

patients = pd.read_csv(
    CLEAN_PATH / "patients_cleaned.csv"
)

print("Patients dataset loaded successfully!")
print()


# ------------------------------------------------------------
# 2. BASIC INFORMATION
# ------------------------------------------------------------

print("Number of rows and columns:")
print(patients.shape)

print()

print("Column names:")
print(patients.columns.tolist())

print()

print("First 5 rows:")
display(patients.head())


# ------------------------------------------------------------
# 3. CHECK PATIENT IDs
# ------------------------------------------------------------

print()
print("Checking Patient IDs...")

duplicate_patient_ids = patients["Id"].duplicated().sum()
missing_patient_ids = patients["Id"].isna().sum()

print("Duplicate Patient IDs:", duplicate_patient_ids)
print("Missing Patient IDs:", missing_patient_ids)


# ------------------------------------------------------------
# 4. CHECK WHETHER ALL ENCOUNTER PATIENT IDs
#    EXIST IN THE PATIENTS DATASET
# ------------------------------------------------------------

encounter_patient_ids = set(encounters["PATIENT"].dropna())
patient_ids = set(patients["Id"].dropna())

missing_patient_references = encounter_patient_ids - patient_ids

print()
print(
    "Patient IDs in encounters but missing from patients:",
    len(missing_patient_references)
)


# ------------------------------------------------------------
# 5. FINAL VALIDATION
# ------------------------------------------------------------

print()
print("Patients dataset validation completed.")
```

```python
# ============================================================
# STEP 4 — PREPARE PATIENT INFORMATION (CORRECTED)
# ============================================================

# ------------------------------------------------------------
# 1. SELECT ONLY THE PATIENT COLUMNS WE NEED
# ------------------------------------------------------------

patient_features = patients[
    [
        "Id",
        "BIRTHDATE",
        "GENDER",
        "RACE",
        "ETHNICITY",
        "MARITAL"
    ]
].copy()


# ------------------------------------------------------------
# 2. CONVERT DATES
# ------------------------------------------------------------

patient_features["BIRTHDATE"] = pd.to_datetime(
    patient_features["BIRTHDATE"],
    errors="coerce",
    utc=True
)

encounters["START"] = pd.to_datetime(
    encounters["START"],
    errors="coerce",
    utc=True
)


# ------------------------------------------------------------
# 3. MERGE PATIENT INFORMATION WITH ENCOUNTERS
# ------------------------------------------------------------

master = encounters.merge(
    patient_features,
    left_on="PATIENT",
    right_on="Id",
    how="left",
    suffixes=("", "_PATIENT")
)


# ------------------------------------------------------------
# 4. CALCULATE AGE AT THE TIME OF ENCOUNTER
# ------------------------------------------------------------

master["AGE_AT_ENCOUNTER"] = (
    (master["START"] - master["BIRTHDATE"]).dt.days / 365.25
).round(1)


# ------------------------------------------------------------
# 5. REMOVE THE PATIENT TABLE ID
# ------------------------------------------------------------

master.drop(
    columns=["Id_PATIENT"],
    inplace=True,
    errors="ignore"
)


# ------------------------------------------------------------
# 6. CHECK THE RESULT
# ------------------------------------------------------------

print("Patient information added successfully!")
print()

print("Master dataset shape:", master.shape)

print()

print("Master dataset columns:")
print(master.columns.tolist())

print()

print("Patient-related information:")
display(
    master[
        [
            "PATIENT",
            "START",
            "TOTAL_CLAIM_COST",
            "AGE_AT_ENCOUNTER",
            "GENDER",
            "RACE",
            "ETHNICITY",
            "MARITAL"
        ]
    ].head()
)
```

```python
# ============================================================
# STEP 5 — VALIDATE PATIENT INFORMATION MERGE
# ============================================================

print("Validating patient information in the master dataset...")
print()


# ------------------------------------------------------------
# 1. CHECK MASTER ROW COUNT
# ------------------------------------------------------------

print("Master rows:", len(master))
print("Original encounter rows:", len(encounters))


# ------------------------------------------------------------
# 2. CHECK DUPLICATE ENCOUNTER IDs
# ------------------------------------------------------------

duplicate_master_ids = master["Id"].duplicated().sum()

print()
print("Duplicate Encounter IDs in master:", duplicate_master_ids)


# ------------------------------------------------------------
# 3. CHECK MISSING PATIENT INFORMATION
# ------------------------------------------------------------

missing_gender = master["GENDER"].isna().sum()
missing_race = master["RACE"].isna().sum()
missing_ethnicity = master["ETHNICITY"].isna().sum()
missing_birthdate = master["BIRTHDATE"].isna().sum()

print()
print("Missing GENDER:", missing_gender)
print("Missing RACE:", missing_race)
print("Missing ETHNICITY:", missing_ethnicity)
print("Missing BIRTHDATE:", missing_birthdate)


# ------------------------------------------------------------
# 4. CHECK AGE VALUES
# ------------------------------------------------------------

negative_age = (master["AGE_AT_ENCOUNTER"] < 0).sum()

print()
print("Negative AGE_AT_ENCOUNTER:", negative_age)


# ------------------------------------------------------------
# 5. CHECK AGE SUMMARY
# ------------------------------------------------------------

print()
print("Age summary:")

display(
    master["AGE_AT_ENCOUNTER"].describe()
)


# ------------------------------------------------------------
# 6. FINAL CHECK
# ------------------------------------------------------------

print()
print("Patient merge validation completed.")
```

```python
from pathlib import Path
import pandas as pd

PROJECT_PATH = Path(
    r"C:\Users\vicky\OneDrive\Desktop\PROJECTS\Healthcare_Cost_Optimization_Analytics"
)

CLEAN_PATH = PROJECT_PATH / "data" / "Cleaned_Datasets"
ANALYTICAL_PATH = PROJECT_PATH / "data" / "Analytical_Dataset"

print("Setup completed successfully!")
```

```python
procedures = pd.read_csv(
    CLEAN_PATH / "procedures_cleaned.csv"
)

print("Procedures dataset loaded successfully!")
print()
print("Shape:")
print(procedures.shape)
print()
print("Columns:")
print(procedures.columns.tolist())
print()
print("First 5 rows:")
display(procedures.head())
```

```python
# Load Encounters
encounters = pd.read_csv(
    CLEAN_PATH / "encounters_cleaned.csv"
)

# Load Patients
patients = pd.read_csv(
    CLEAN_PATH / "patients_cleaned.csv"
)

# Select patient features
patient_features = patients[
    ["Id", "BIRTHDATE", "GENDER", "RACE", "ETHNICITY", "MARITAL"]
].copy()

# Convert dates
patient_features["BIRTHDATE"] = pd.to_datetime(
    patient_features["BIRTHDATE"],
    errors="coerce",
    utc=True
)

encounters["START"] = pd.to_datetime(
    encounters["START"],
    errors="coerce",
    utc=True
)

# Merge patient information into encounters
master = encounters.merge(
    patient_features,
    left_on="PATIENT",
    right_on="Id",
    how="left",
    suffixes=("", "_PATIENT")
)

# Calculate age at encounter
master["AGE_AT_ENCOUNTER"] = (
    (master["START"] - master["BIRTHDATE"]).dt.days / 365.25
).round(1)

# Remove duplicate patient ID column
master.drop(
    columns=["Id_PATIENT"],
    inplace=True,
    errors="ignore"
)

print("Base master recreated successfully!")
print("Master shape:", master.shape)
```

```python
# Load Conditions dataset again

conditions = pd.read_csv(
    CLEAN_PATH / "conditions_cleaned.csv"
)

print("Conditions dataset loaded successfully!")
print("Shape:", conditions.shape)
```

```python
condition_summary = (
    conditions
    .groupby("ENCOUNTER")
    .size()
    .reset_index(name="condition_count")
)

print("Condition summary recreated successfully!")
print("Shape:", condition_summary.shape)
```

```python
master = master.merge(
    condition_summary,
    left_on="Id",
    right_on="ENCOUNTER",
    how="left"
)

# Condition record illaadha encounters-ku 0
master["condition_count"] = (
    master["condition_count"]
    .fillna(0)
    .astype(int)
)

# Extra ENCOUNTER column remove pannuvom
master.drop(
    columns=["ENCOUNTER"],
    inplace=True
)

print("Conditions merged into master successfully!")
print()
print("Master shape:", master.shape)
```

```python
print("Master rows:", len(master))
print("Duplicate Encounter IDs:", master["Id"].duplicated().sum())
print("Missing condition_count:", master["condition_count"].isna().sum())
print("Minimum condition_count:", master["condition_count"].min())
print("Maximum condition_count:", master["condition_count"].max())
```

```python
procedure_encounters = set(
    procedures["ENCOUNTER"].dropna()
)

master_encounters = set(
    master["Id"].dropna()
)

invalid_procedure_encounters = (
    procedure_encounters - master_encounters
)

print("Unique Encounter IDs in Procedures:", len(procedure_encounters))
print("Unique Encounter IDs in Master:", len(master_encounters))
print(
    "Procedure Encounter IDs not found in Master:",
    len(invalid_procedure_encounters)
)
```

```python
procedure_summary = (
    procedures
    .groupby("ENCOUNTER")
    .agg(
        procedure_count=("CODE", "count"),
        procedure_cost=("BASE_COST", "sum")
    )
    .reset_index()
)

print("Procedure summary created successfully!")
print()
print("Shape:", procedure_summary.shape)
print()
print("First 5 rows:")
display(procedure_summary.head())
```

```python
master = master.merge(
    procedure_summary,
    left_on="Id",
    right_on="ENCOUNTER",
    how="left"
)

# Procedure record illaadha encounters-ku 0
master["procedure_count"] = (
    master["procedure_count"]
    .fillna(0)
    .astype(int)
)

master["procedure_cost"] = (
    master["procedure_cost"]
    .fillna(0)
)

# Extra ENCOUNTER column remove pannuvom
master.drop(
    columns=["ENCOUNTER"],
    inplace=True
)

print("Procedures merged into master successfully!")
print()
print("Master shape:", master.shape)
```

```python
print("Master rows:", len(master))
print("Duplicate Encounter IDs:", master["Id"].duplicated().sum())
print("Missing procedure_count:", master["procedure_count"].isna().sum())
print("Missing procedure_cost:", master["procedure_cost"].isna().sum())
print("Minimum procedure_count:", master["procedure_count"].min())
print("Maximum procedure_count:", master["procedure_count"].max())
print("Minimum procedure_cost:", master["procedure_cost"].min())
print("Maximum procedure_cost:", master["procedure_cost"].max())
```

```python
# Validate Claims Transactions -> Encounters relationship

transaction_appointments = (
    claims_transactions["APPOINTMENTID"]
    .dropna()
    .unique()
)

master_encounters = master["Id"].unique()

print("Unique transaction appointment IDs:",
      len(transaction_appointments))

print("Unique master encounter IDs:",
      len(master_encounters))

print(
    "Transaction appointment IDs not in master:",
    (~claims_transactions["APPOINTMENTID"].isin(master_encounters)).sum()
)

print(
    "Master encounters with no transaction records:",
    (~master["Id"].isin(transaction_appointments)).sum()
)

print(
    "Missing APPOINTMENTID in transactions:",
    claims_transactions["APPOINTMENTID"].isna().sum()
)
```

```python
transaction_summary = (
    claims_transactions
    .groupby("APPOINTMENTID")
    .agg(
        charge_amount=("AMOUNT", lambda x: x[claims_transactions.loc[x.index, "TYPE"].eq("CHARGE")].sum()),
        payment_amount=("PAYMENTS", "sum"),
        adjustment_amount=("ADJUSTMENTS", "sum"),
        outstanding_amount=("OUTSTANDING", "sum"),
        transaction_count=("ID", "count")
    )
    .reset_index()
)

print("Transaction summary shape:", transaction_summary.shape)

print("\nColumns:")
print(transaction_summary.columns.tolist())

print("\nFirst 5 rows:")
display(transaction_summary.head())
```

```python
# Create encounter-level transaction summary

transaction_summary = (
    claims_transactions
    .groupby("APPOINTMENTID")
    .agg(
        transaction_count=("ID", "count"),
        total_amount=("AMOUNT", "sum"),
        total_payments=("PAYMENTS", "sum"),
        total_adjustments=("ADJUSTMENTS", "sum"),
        total_outstanding=("OUTSTANDING", "sum")
    )
    .reset_index()
)

print("Transaction summary shape:", transaction_summary.shape)

print("\nColumns:")
print(transaction_summary.columns.tolist())

print("\nFirst 5 rows:")
display(transaction_summary.head())
```

```python
# Create the final clean encounter-level transaction summary

transaction_summary = (
    claims_transactions
    .groupby("APPOINTMENTID")
    .agg(
        charge_amount=(
            "AMOUNT",
            lambda x: x[
                claims_transactions.loc[x.index, "TYPE"].eq("CHARGE")
            ].sum()
        ),
        payment_amount=("PAYMENTS", "sum"),
        adjustment_amount=("ADJUSTMENTS", "sum"),
        outstanding_amount=("OUTSTANDING", "sum"),
        transaction_count=("ID", "count")
    )
    .reset_index()
)

print("Transaction summary shape:", transaction_summary.shape)

print("\nColumns:")
print(transaction_summary.columns.tolist())

print("\nFirst 5 rows:")
display(transaction_summary.head())
```

```python
print("Financial summary:")

print(
    transaction_summary[
        [
            "charge_amount",
            "payment_amount",
            "adjustment_amount",
            "outstanding_amount",
            "transaction_count"
        ]
    ].describe()
)

print("\nMissing values:")
print(
    transaction_summary[
        [
            "charge_amount",
            "payment_amount",
            "adjustment_amount",
            "outstanding_amount"
        ]
    ].isna().sum()
)

print("\nNegative values:")
print(
    (
        transaction_summary[
            [
                "charge_amount",
                "payment_amount",
                "adjustment_amount",
                "outstanding_amount"
            ]
        ] < 0
    ).sum()
)
```

```python
print("Transaction summary rows:", len(transaction_summary))

print(
    "Duplicate APPOINTMENTID:",
    transaction_summary["APPOINTMENTID"].duplicated().sum()
)

print(
    "APPOINTMENTIDs not in master:",
    (~transaction_summary["APPOINTMENTID"].isin(master["Id"])).sum()
)

print(
    "Master encounters not in transaction summary:",
    (~master["Id"].isin(transaction_summary["APPOINTMENTID"])).sum()
)
```

```python
# Integrate Claims Transactions features into the master dataset

master = master.merge(
    transaction_summary[
        [
            "APPOINTMENTID",
            "charge_amount",
            "payment_amount",
            "outstanding_amount",
            "transaction_count"
        ]
    ],
    how="left",
    left_on="Id",
    right_on="APPOINTMENTID"
)

# Rename transaction features
master.rename(
    columns={
        "charge_amount": "transaction_charge_amount",
        "payment_amount": "transaction_payment_amount",
        "outstanding_amount": "transaction_outstanding_amount",
        "transaction_count": "transaction_count"
    },
    inplace=True
)

# Remove the temporary join key
master.drop(columns=["APPOINTMENTID"], inplace=True)

print("Master shape:", master.shape)

print("\nTransaction-related columns:")
print([
    col for col in master.columns
    if "transaction" in col.lower()
])

print("\nDuplicate Encounter IDs:",
      master["Id"].duplicated().sum())

print("\nMissing transaction features:")
print(
    master[
        [
            "transaction_charge_amount",
            "transaction_payment_amount",
            "transaction_outstanding_amount",
            "transaction_count"
        ]
    ].isna().sum()
)
```

```python
payer_check = payer_check.sort_values(
    ["PATIENT", "START_YEAR"]
)

encounter_check = encounter_check.sort_values(
    ["PATIENT", "START"]
)

payer_check.head()
```

```python
encounter_payer_match = encounter_check.merge(
    payer_check,
    on="PATIENT",
    how="left"
)

encounter_payer_match["payer_period_match"] = (
    (encounter_payer_match["START"] >= encounter_payer_match["START_YEAR"]) &
    (encounter_payer_match["START"] <= encounter_payer_match["END_YEAR"])
)

print("Total encounter-payer combinations:",
      len(encounter_payer_match))

print("\nMatched combinations:",
      encounter_payer_match["payer_period_match"].sum())

print("\nEncounters with at least one matching payer period:",
      encounter_payer_match
      .groupby("Id")["payer_period_match"]
      .any()
      .sum())

print("\nEncounters with NO matching payer period:",
      (~encounter_payer_match
       .groupby("Id")["payer_period_match"]
       .any()).sum())
```

```python
match_counts = (
    encounter_payer_match
    .groupby("Id")["payer_period_match"]
    .sum()
)

print("Encounters with 0 matches:", (match_counts == 0).sum())
print("Encounters with 1 match:", (match_counts == 1).sum())
print("Encounters with 2 matches:", (match_counts == 2).sum())
print("Encounters with more than 2 matches:", (match_counts > 2).sum())
print("Maximum matches for one encounter:", match_counts.max())
```

```python
double_matches = encounter_payer_match[
    encounter_payer_match["payer_period_match"]
].copy()

double_match_ids = (
    double_matches.groupby("Id")
    .size()
)

double_match_ids = double_match_ids[
    double_match_ids == 2
].index

double_match_examples = double_matches[
    double_matches["Id"].isin(double_match_ids)
].sort_values(["Id", "START_YEAR"])

double_match_examples[
    ["Id", "PATIENT", "START", "START_YEAR", "END_YEAR", "PAYER"]
].head(20)
```
