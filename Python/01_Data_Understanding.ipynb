```python
# Phase 2 - Data Cleaning & Validation
# Step 2.2 - Load core datasets

import pandas as pd

patients = pd.read_csv("../data/Raw_Datasets/patients.csv")
encounters = pd.read_csv("../data/Raw_Datasets/encounters.csv")
providers = pd.read_csv("../data/Raw_Datasets/providers.csv")
organizations = pd.read_csv("../data/Raw_Datasets/organizations.csv")
payers = pd.read_csv("../data/Raw_Datasets/payers.csv")
conditions = pd.read_csv("../data/Raw_Datasets/conditions.csv")
procedures = pd.read_csv("../data/Raw_Datasets/procedures.csv")

print("Core datasets loaded successfully.")
```

```python
# Check: Do all encounter patient IDs exist in patients table?

patient_ids_in_patients = set(patients["Id"])
patient_ids_in_encounters = set(encounters["PATIENT"])

missing_patient_ids = patient_ids_in_encounters - patient_ids_in_patients

print("Unique patient IDs in encounters:", len(patient_ids_in_encounters))
print("Patient IDs in patients table:", len(patient_ids_in_patients))
print("Patient IDs in encounters not found in patients:", len(missing_patient_ids))
```

```python
# Check: Do all encounter provider IDs exist in providers table?

provider_ids_in_providers = set(providers["Id"])
provider_ids_in_encounters = set(encounters["PROVIDER"])

missing_provider_ids = provider_ids_in_encounters - provider_ids_in_providers

print("Unique provider IDs in encounters:", len(provider_ids_in_encounters))
print("Provider IDs in providers table:", len(provider_ids_in_providers))
print("Provider IDs in encounters not found in providers:", len(missing_provider_ids))
```

```python
# Check: Do all encounter organization IDs exist in organizations table?

organization_ids_in_organizations = set(organizations["Id"])
organization_ids_in_encounters = set(encounters["ORGANIZATION"])

missing_organization_ids = (
    organization_ids_in_encounters - organization_ids_in_organizations
)

print("Unique organization IDs in encounters:", len(organization_ids_in_encounters))
print("Organization IDs in organizations table:", len(organization_ids_in_organizations))
print(
    "Organization IDs in encounters not found in organizations:",
    len(missing_organization_ids)
)
```

```python
# Check: Do all encounter payer IDs exist in payers table?

payer_ids_in_payers = set(payers["Id"])
payer_ids_in_encounters = set(encounters["PAYER"])

missing_payer_ids = payer_ids_in_encounters - payer_ids_in_payers

print("Unique payer IDs in encounters:", len(payer_ids_in_encounters))
print("Payer IDs in payers table:", len(payer_ids_in_payers))
print("Payer IDs in encounters not found in payers:", len(missing_payer_ids))
```

```python
# Check: Do all condition encounter IDs exist in encounters table?

encounter_ids_in_encounters = set(encounters["Id"])
encounter_ids_in_conditions = set(conditions["ENCOUNTER"])

missing_condition_encounters = (
    encounter_ids_in_conditions - encounter_ids_in_encounters
)

print("Unique encounter IDs in conditions:", len(encounter_ids_in_conditions))
print("Encounter IDs in encounters table:", len(encounter_ids_in_encounters))
print(
    "Condition encounter IDs not found in encounters:",
    len(missing_condition_encounters)
)
```

```python
# Check: Do all procedure encounter IDs exist in encounters table?

encounter_ids_in_procedures = set(procedures["ENCOUNTER"])

missing_procedure_encounters = (
    encounter_ids_in_procedures - encounter_ids_in_encounters
)

print("Unique encounter IDs in procedures:", len(encounter_ids_in_procedures))
print("Encounter IDs in encounters table:", len(encounter_ids_in_encounters))
print(
    "Procedure encounter IDs not found in encounters:",
    len(missing_procedure_encounters)
)
```

```python
# Step 2.6 - Calculate encounter duration

encounters["ENCOUNTER_DURATION_HOURS"] = (
    encounters["STOP"] - encounters["START"]
).dt.total_seconds() / 3600

print(encounters["ENCOUNTER_DURATION_HOURS"].describe())
```

```python
# Find encounters with unusually long durations

long_encounters = encounters[
    encounters["ENCOUNTER_DURATION_HOURS"] > 24 * 30
][[
    "Id",
    "PATIENT",
    "START",
    "STOP",
    "ENCOUNTERCLASS",
    "DESCRIPTION",
    "ENCOUNTER_DURATION_HOURS"
]]

print("Encounters longer than 30 days:", len(long_encounters))

long_encounters.head(10)
```

```python
# Check long-duration encounters by encounter type

long_duration_summary = (
    encounters[encounters["ENCOUNTER_DURATION_HOURS"] > 24 * 30]
    .groupby("ENCOUNTERCLASS")
    .size()
    .sort_values(ascending=False)
)

print(long_duration_summary)
```

```python
# ============================================================
# REMAINING DATASETS - COMPLETE VALIDATION
# Healthcare Cost Optimization Analytics
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np

# ------------------------------------------------------------
# 1. DATA PATH
# ------------------------------------------------------------

DATA_PATH = Path("../data/Raw_Datasets")

datasets_to_validate = [
    "claims_transactions.csv",
    "medications.csv",
    "observations.csv",
    "careplans.csv",
    "imaging_studies.csv",
    "allergies.csv",
    "immunizations.csv",
    "payer_transitions.csv"
]

# ------------------------------------------------------------
# 2. LOAD ALL DATASETS
# ------------------------------------------------------------

data = {}

for file_name in datasets_to_validate:
    file_path = DATA_PATH / file_name

    df = pd.read_csv(file_path)
    data[file_name] = df

    print("=" * 80)
    print(f"{file_name}")
    print("=" * 80)
    print(f"Rows    : {df.shape[0]:,}")
    print(f"Columns : {df.shape[1]}")
    print()
    print("Columns:")
    print(df.columns.tolist())
    print()
    print("Data types:")
    print(df.dtypes)
    print()


# ============================================================
# 3. GENERAL STRUCTURAL VALIDATION
# ============================================================

for file_name, df in data.items():

    print("\n" + "=" * 80)
    print(f"STRUCTURAL VALIDATION - {file_name}")
    print("=" * 80)

    print(f"Total rows: {len(df):,}")
    print(f"Total columns: {len(df.columns)}")

    # --------------------------------------------------------
    # Duplicate rows
    # --------------------------------------------------------

    duplicate_rows = df.duplicated().sum()

    print(f"Duplicate complete rows: {duplicate_rows:,}")

    # --------------------------------------------------------
    # ID validation
    # --------------------------------------------------------

    if "Id" in df.columns:

        duplicate_ids = df["Id"].duplicated().sum()
        missing_ids = df["Id"].isna().sum()
        unique_ids = df["Id"].nunique(dropna=True)

        print(f"Unique IDs: {unique_ids:,}")
        print(f"Duplicate IDs: {duplicate_ids:,}")
        print(f"Missing IDs: {missing_ids:,}")

    # --------------------------------------------------------
    # Missing values
    # --------------------------------------------------------

    missing = df.isna().sum()
    missing = missing[missing > 0].sort_values(ascending=False)

    print("\nMissing values:")

    if len(missing) == 0:
        print("No missing values.")
    else:
        print(missing)

    # --------------------------------------------------------
    # Numeric columns
    # --------------------------------------------------------

    numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

    if numeric_cols:

        print("\nNumeric columns:")

        for col in numeric_cols:

            negative_count = (df[col] < 0).sum()
            zero_count = (df[col] == 0).sum()

            print(
                f"{col} -> "
                f"negative: {negative_count:,} | "
                f"zero: {zero_count:,}"
            )


# ============================================================
# 4. PATIENT REFERENCE VALIDATION
# ============================================================

print("\n" + "=" * 80)
print("PATIENT REFERENCE VALIDATION")
print("=" * 80)

patients = data.get("patients.csv")

# patients.csv was already loaded earlier in the notebook.
# If it is not currently available, load it safely.

if patients is None:
    patients = pd.read_csv(DATA_PATH / "patients.csv")

patient_ids = set(patients["Id"].dropna())

for file_name, df in data.items():

    patient_columns = [
        col for col in df.columns
        if col.upper() in [
            "PATIENT",
            "PATIENTID",
            "PATIENT_ID"
        ]
    ]

    for col in patient_columns:

        source_ids = set(df[col].dropna())

        missing_refs = source_ids - patient_ids

        print(f"\n{file_name} -> {col}")
        print(f"Unique patient references: {len(source_ids):,}")
        print(f"Missing patient references: {len(missing_refs):,}")

        if len(missing_refs) == 0:
            print("✓ All patient references are valid.")
        else:
            print("⚠ Some patient references are not present in patients.csv.")


# ============================================================
# 5. PROVIDER REFERENCE VALIDATION
# ============================================================

print("\n" + "=" * 80)
print("PROVIDER REFERENCE VALIDATION")
print("=" * 80)

providers = data.get("providers.csv")

if providers is None:
    providers = pd.read_csv(DATA_PATH / "providers.csv")

provider_ids = set(providers["Id"].dropna())

for file_name, df in data.items():

    provider_columns = [
        col for col in df.columns
        if "PROVIDER" in col.upper()
    ]

    for col in provider_columns:

        source_ids = set(df[col].dropna())

        missing_refs = source_ids - provider_ids

        print(f"\n{file_name} -> {col}")
        print(f"Unique provider references: {len(source_ids):,}")
        print(f"Missing provider references: {len(missing_refs):,}")

        if len(missing_refs) == 0:
            print("✓ All provider references are valid.")
        else:
            print("⚠ Some provider references are not present in providers.csv.")


# ============================================================
# 6. ENCOUNTER REFERENCE VALIDATION
# ============================================================

print("\n" + "=" * 80)
print("ENCOUNTER REFERENCE VALIDATION")
print("=" * 80)

encounters = data.get("encounters.csv")

if encounters is None:
    encounters = pd.read_csv(DATA_PATH / "encounters.csv")

encounter_ids = set(encounters["Id"].dropna())

for file_name, df in data.items():

    encounter_columns = [
        col for col in df.columns
        if col.upper() in [
            "ENCOUNTER",
            "ENCOUNTERID",
            "ENCOUNTER_ID"
        ]
    ]

    for col in encounter_columns:

        source_ids = set(df[col].dropna())

        missing_refs = source_ids - encounter_ids

        print(f"\n{file_name} -> {col}")
        print(f"Unique encounter references: {len(source_ids):,}")
        print(f"Missing encounter references: {len(missing_refs):,}")

        if len(missing_refs) == 0:
            print("✓ All encounter references are valid.")
        else:
            print("⚠ Some encounter references are not present in encounters.csv.")


# ============================================================
# 7. PAYER REFERENCE VALIDATION
# ============================================================

print("\n" + "=" * 80)
print("PAYER REFERENCE VALIDATION")
print("=" * 80)

payers = data.get("payers.csv")

if payers is None:
    payers = pd.read_csv(DATA_PATH / "payers.csv")

payer_ids = set(payers["Id"].dropna())

for file_name, df in data.items():

    payer_columns = [
        col for col in df.columns
        if "PAYER" in col.upper()
    ]

    for col in payer_columns:

        source_ids = set(df[col].dropna())

        missing_refs = source_ids - payer_ids

        print(f"\n{file_name} -> {col}")
        print(f"Unique payer references: {len(source_ids):,}")
        print(f"Missing payer references: {len(missing_refs):,}")

        if len(missing_refs) == 0:
            print("✓ All payer references are valid.")
        else:
            print("⚠ Some payer references are not present in payers.csv.")


# ============================================================
# 8. DATE COLUMN VALIDATION
# ============================================================

print("\n" + "=" * 80)
print("DATE COLUMN VALIDATION")
print("=" * 80)

for file_name, df in data.items():

    print(f"\n--- {file_name} ---")

    date_columns = []

    for col in df.columns:

        col_upper = col.upper()

        # Identify likely date columns
        if any(keyword in col_upper for keyword in [
            "DATE",
            "START",
            "STOP",
            "TIME"
        ]):

            date_columns.append(col)

    if not date_columns:
        print("No obvious date columns detected.")
        continue

    for col in date_columns:

        converted = pd.to_datetime(
            df[col],
            errors="coerce",
            utc=True
        )

        invalid = (
            df[col].notna() &
            converted.isna()
        ).sum()

        missing = converted.isna().sum()

        print(
            f"{col} -> "
            f"Invalid values: {invalid:,} | "
            f"Missing after conversion: {missing:,}"
        )


# ============================================================
# 9. NEGATIVE NUMERIC VALUE CHECK
# ============================================================

print("\n" + "=" * 80)
print("NEGATIVE VALUE CHECK")
print("=" * 80)

for file_name, df in data.items():

    numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

    if not numeric_cols:
        continue

    print(f"\n--- {file_name} ---")

    found_negative = False

    for col in numeric_cols:

        negative_count = (df[col] < 0).sum()

        if negative_count > 0:

            found_negative = True

            print(
                f"{col} -> "
                f"Negative values: {negative_count:,}"
            )

    if not found_negative:
        print("No negative numeric values found.")


# ============================================================
# 10. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("REMAINING DATASETS VALIDATION SUMMARY")
print("=" * 80)

summary_rows = []

for file_name, df in data.items():

    summary_rows.append({
        "Dataset": file_name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Duplicate Rows": df.duplicated().sum(),
        "Missing Cells": int(df.isna().sum().sum())
    })

validation_summary = pd.DataFrame(summary_rows)

print(validation_summary.to_string(index=False))

print("\n" + "=" * 80)
print("ALL REMAINING DATASET VALIDATIONS COMPLETED")
print("=" * 80)
```

```python
# ============================================================
# FINAL TARGETED VALIDATION
# Observations + Imaging + Claims Transactions
# ============================================================

# ============================================================
# 1. OBSERVATIONS - INVESTIGATE EXACT DUPLICATES
# ============================================================

print("=" * 80)
print("OBSERVATIONS - DUPLICATE INVESTIGATION")
print("=" * 80)

observations = data["observations.csv"]

duplicate_observations = observations[
    observations.duplicated(keep=False)
].sort_values(
    by=observations.columns.tolist()
)

print(f"Total exact duplicate rows: {observations.duplicated().sum():,}")
print(f"Rows involved in exact duplicates: {len(duplicate_observations):,}")

print("\nSample duplicate observations:")
print(duplicate_observations.head(20).to_string(index=False))


# ============================================================
# 2. IMAGING STUDIES - IDENTIFIER INVESTIGATION
# ============================================================

print("\n" + "=" * 80)
print("IMAGING STUDIES - IDENTIFIER VALIDATION")
print("=" * 80)

imaging = data["imaging_studies.csv"]

for col in [
    "Id",
    "SERIES_UID",
    "INSTANCE_UID",
    "ENCOUNTER",
    "PATIENT"
]:

    if col in imaging.columns:

        print(
            f"{col}: "
            f"Rows={len(imaging):,} | "
            f"Unique={imaging[col].nunique(dropna=True):,} | "
            f"Missing={imaging[col].isna().sum():,}"
        )

print("\nDuplicate counts:")

for col in [
    "Id",
    "SERIES_UID",
    "INSTANCE_UID"
]:

    if col in imaging.columns:

        print(
            f"{col} duplicate rows: "
            f"{imaging[col].duplicated().sum():,}"
        )


# ============================================================
# 3. CLAIMS TRANSACTIONS - FINANCIAL VALIDATION
# ============================================================

print("\n" + "=" * 80)
print("CLAIMS TRANSACTIONS - FINANCIAL VALIDATION")
print("=" * 80)

transactions = data["claims_transactions.csv"]

financial_columns = [
    "AMOUNT",
    "UNITAMOUNT",
    "PAYMENTS",
    "ADJUSTMENTS",
    "TRANSFERS",
    "OUTSTANDING"
]

existing_financial_columns = [
    col for col in financial_columns
    if col in transactions.columns
]

print("Financial columns:")
print(existing_financial_columns)

print("\nFinancial summary:")

print(
    transactions[existing_financial_columns]
    .describe()
    .T
)

print("\nNegative values:")

for col in existing_financial_columns:

    negative = (transactions[col] < 0).sum()

    print(
        f"{col} -> negative values: {negative:,}"
    )


# ============================================================
# 4. CLAIMS TRANSACTIONS - DATE LOGIC
# ============================================================

print("\n" + "=" * 80)
print("CLAIMS TRANSACTIONS - DATE LOGIC")
print("=" * 80)

from_date = pd.to_datetime(
    transactions["FROMDATE"],
    errors="coerce",
    utc=True
)

to_date = pd.to_datetime(
    transactions["TODATE"],
    errors="coerce",
    utc=True
)

invalid_date_order = (
    (to_date < from_date)
).sum()

print(
    f"TODATE before FROMDATE: "
    f"{invalid_date_order:,}"
)


# ============================================================
# 5. CLAIM TRANSACTION / CLAIM REFERENCE CHECK
# ============================================================

print("\n" + "=" * 80)
print("CLAIM TRANSACTION -> CLAIM REFERENCE")
print("=" * 80)

claims = pd.read_csv(DATA_PATH / "claims.csv")

claim_ids = set(claims["Id"].dropna())

transaction_claim_ids = set(
    transactions["CLAIMID"].dropna()
)

missing_claim_refs = (
    transaction_claim_ids - claim_ids
)

print(
    f"Unique CLAIMID references: "
    f"{len(transaction_claim_ids):,}"
)

print(
    f"Missing CLAIMID references: "
    f"{len(missing_claim_refs):,}"
)

if len(missing_claim_refs) == 0:
    print("✓ All transaction claim references are valid.")
else:
    print(
        "⚠ Some transaction claim references "
        "are not present in claims.csv."
    )


# ============================================================
# 6. FINAL
# ============================================================

print("\n" + "=" * 80)
print("TARGETED VALIDATION COMPLETED")
print("=" * 80)
```
