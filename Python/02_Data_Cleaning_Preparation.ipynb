```python
import pandas as pd

# Load the raw encounters dataset
encounters = pd.read_csv(RAW_PATH / "encounters.csv")

# Check the dataset
print("Shape:", encounters.shape)
print("\nColumns:")
print(encounters.columns.tolist())
```

```python
# Check encounter class values and their frequencies
encounter_class_counts = encounters["ENCOUNTERCLASS"].value_counts()

print(encounter_class_counts)
```

```python
# Check important ID/reference columns
reference_columns = [
    "PATIENT",
    "ORGANIZATION",
    "PROVIDER",
    "PAYER"
]

for column in reference_columns:
    print(f"{column}: {encounters[column].isna().sum()} missing")
    
```

```python
# Final validation of the cleaned encounters dataset

print("Final shape:", encounters.shape)
print("Duplicate rows:", encounters.duplicated().sum())
print("Duplicate encounter IDs:", encounters["Id"].duplicated().sum())
print("Missing values:", encounters.isna().sum().sum())
```

```python
# ============================================================
# STEP 4: CONVERT AND VALIDATE DATES
# ============================================================

# Convert START and STOP to datetime
conditions["START"] = pd.to_datetime(conditions["START"], errors="coerce")
conditions["STOP"] = pd.to_datetime(conditions["STOP"], errors="coerce")

print("Date conversion completed.")

print("\nData types:")
print(conditions[["START", "STOP"]].dtypes)

# Check invalid date conversion
print("\nInvalid START dates:", conditions["START"].isna().sum())
print("Invalid STOP dates:", conditions["STOP"].isna().sum())

# Check whether STOP occurs before START
invalid_date_order = (
    conditions["STOP"].notna() &
    (conditions["STOP"] < conditions["START"])
).sum()

print("\nSTOP dates before START dates:", invalid_date_order)
```

```python
# ============================================================
# STEP 5: CHECK CONDITION CODE AND DESCRIPTION
# ============================================================

print("Missing CODE values:", conditions["CODE"].isna().sum())
print("Missing DESCRIPTION values:", conditions["DESCRIPTION"].isna().sum())

print("\nUnique condition codes:", conditions["CODE"].nunique())
print("Unique condition descriptions:", conditions["DESCRIPTION"].nunique())

print("\nSample condition records:")
print(conditions[["CODE", "DESCRIPTION"]].head(10))
```

```python
# ============================================================
# STEP 6: CHECK PATIENT AND ENCOUNTER REFERENCES
# ============================================================

# Get valid IDs from cleaned datasets
valid_patient_ids = set(patients["Id"])
valid_encounter_ids = set(encounters["Id"])

# Find condition records with invalid patient references
invalid_patient_refs = (~conditions["PATIENT"].isin(valid_patient_ids)).sum()

# Find condition records with invalid encounter references
invalid_encounter_refs = (~conditions["ENCOUNTER"].isin(valid_encounter_ids)).sum()

print("Invalid PATIENT references:", invalid_patient_refs)
print("Invalid ENCOUNTER references:", invalid_encounter_refs)

# Show number of unique patients and encounters in conditions
print("\nUnique patients in conditions:", conditions["PATIENT"].nunique())
print("Unique encounters in conditions:", conditions["ENCOUNTER"].nunique())
```

```python
# ============================================================
# STEP 7: CHECK CONDITIONS PER ENCOUNTER
# ============================================================

conditions_per_encounter = (
    conditions.groupby("ENCOUNTER")
    .size()
)

print("Conditions per encounter:")
print(conditions_per_encounter.describe())

print("\nMinimum conditions for an encounter:",
      conditions_per_encounter.min())

print("Maximum conditions for an encounter:",
      conditions_per_encounter.max())

print("Median conditions per encounter:",
      conditions_per_encounter.median())

print("Average conditions per encounter:",
      conditions_per_encounter.mean())
```

```python
# ============================================================
# STEP 8: FINAL VALIDATION
# ============================================================

print("Final Conditions Dataset Validation")
print("=" * 50)

print("Shape:", conditions.shape)
print("Duplicate rows:", conditions.duplicated().sum())
print("Missing values:")
print(conditions.isna().sum())

print("\nDate validation:")
print("Invalid START dates:", conditions["START"].isna().sum())
print("Invalid STOP before START:",
      (
          conditions["STOP"].notna() &
          (conditions["STOP"] < conditions["START"])
      ).sum())

print("\nReference validation:")
print("Invalid PATIENT references:",
      (~conditions["PATIENT"].isin(valid_patient_ids)).sum())
print("Invalid ENCOUNTER references:",
      (~conditions["ENCOUNTER"].isin(valid_encounter_ids)).sum())

print("\nCondition validation:")
print("Missing CODE:", conditions["CODE"].isna().sum())
print("Missing DESCRIPTION:", conditions["DESCRIPTION"].isna().sum())
print("Unique condition codes:", conditions["CODE"].nunique())

print("\nFinal data types:")
print(conditions.dtypes)
```
